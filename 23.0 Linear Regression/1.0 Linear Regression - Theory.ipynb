### Linear Regression

#### Why do we start with Linear Regression when we study Machine Learning?

- <b>Simplicity:</b> Linear regression is straightforward to understand and implement, making it an excellent entry point for beginners. It involves fitting a line to data points, which is easy to visualize. 📈
- <b>Foundational Concepts:</b> It introduces fundamental concepts such as loss functions, optimization, and the relationship between features and target variables.
- <b>Interpretability:</b> The model is interpretable, allowing us to understand how changes in input features affect the output, which is crucial in many applications.
- <b>Performance Benchmark:</b> Linear regression serves as a baseline model. More complex algorithms can be compared against it to see if they provide significant improvements.
- <b>Wide Applicability:</b> It can be applied to various problems, from predicting sales to understanding relationships between variables, making it versatile.

#### What is Simple Linear Regression?

Simple linear regression is a statistical method (Supervised Machin Learning Algorithm) used to model the relationship between two variables by fitting a straight line to the data. Here's a breakdown of its key components:

- <b>Variables:</b> It involves one independent variable (predictor) and one dependent variable (response). For example, predicting a person's weight based on their height. 📏➡️⚖️
- <b>Equation:</b> The relationship is expressed with the equation: Y = b_0 + b_1X + epsilon
    - Y is the dependent variable,
    - X is the independent variable,
    - b_0 is the y-intercept,
    - b_1 is the slope of the line,
    - epsilon is the error term.
- <b>Objective:</b> The goal is to find the best-fitting line (i.e., the line that minimizes the difference between the predicted values and the actual values), typically using the least squares method.
- <b>Assumptions:</b> It assumes a linear relationship between the variables, that errors are normally distributed, and that there is homoscedasticity (constant variance of errors).

#### What is stochastic error?

Stochastic error, also known as random error, refers to the variability in a dataset that cannot be explained by the model being used. Here are some key points about stochastic error:

- <b>Nature:</b> It arises from random fluctuations or noise in the data, which can be due to various factors like measurement errors, unobserved variables, or inherent randomness in the process being studied.
- <b>Model Representation:</b> In the context of regression models, stochastic error is often represented by the error term (ε) in the regression equation: Y = b_0 + b_1X + ε. Here, ε captures the deviation of the observed values from the predicted values, which is not accounted for by the model.
- <b>Assumption:</b> In many statistical models, it's assumed that stochastic errors are normally distributed with a mean of zero and constant variance (homoscedasticity). This assumption allows for more straightforward statistical inference.
- <b>Impact on Predictions:</b> While stochastic errors introduce uncertainty in predictions, they are a natural part of any real-world data analysis. The goal is to minimize their impact through appropriate modeling techniques.

#### What is Linear and Simple Linear Regression?

Linear regression is a statistical method used to model the relationship between one or more independent variables (predictors) and a dependent variable (response). It aims to find the best-fitting linear equation that describes how the dependent variable changes as the independent variables change. Linear regression can be categorized into two types:

- <b>Simple Linear Regression:</b> Involves one independent variable and one dependent variable.
- <b>Multiple Linear Regression:</b> Involves two or more independent variables predicting a single dependent variable.

- <b>Simple Linear Regression</b><br>
Simple linear regression is a specific case of linear regression that focuses on the relationship between two variables. <br> <b>Here are some key features:</b>
Model Structure: It is represented by the equation: Y = b_0 + b_1X + epsilon
    - Y: Dependent variable (what you are trying to predict)
    - X: Independent variable (the predictor)
    - b_0: Y-intercept (value of Y when X is 0)
    - b_1: Slope of the line (how much Y changes for a one-unit increase in X)
    - epsilon \): Stochastic error (the randomness or noise in the data)
- <b>Purpose:</b> It aims to find the linear relationship between the independent and dependent variables, allowing for predictions of Y based on new values of X.
- <b>Visualization:</b> The relationship is often visualized with a scatter plot, where the best fit line represents the predicted values.
In summary, while linear regression encompasses both simple and multiple linear regression, simple linear regression specifically deals with a single independent variable influencing a dependent variable.

#### What is Best Fit line?

The best fit line, often referred to as the regression line in the context of linear regression, is a straight line that best represents the data points in a scatter plot. Here’s a closer look at what it entails:

- <b>Purpose:</b> The best fit line aims to summarize the relationship between the independent variable (X) and the dependent variable (Y) by providing a simple equation that can be used for prediction. 📈
- <b>Mathematical Representation:</b> The line can be represented by the equation: Y = b_0 + b_1X
    - b_0 is the y-intercept (the value of Y when X is 0),
    - b_1 is the slope of the line (how much Y changes for a one-unit change in X).
- <b>Minimizing Errors:</b> The best fit line is typically determined using the least squares method, which minimizes the sum of the squares of the vertical distances (residuals) between the observed data points and the line itself. This approach helps ensure that the line is as close as possible to all points in the dataset.
- <b>Visual Representation:</b> When plotted on a graph, the best fit line provides a clear visual indication of the trend in the data, making it easier to interpret relationships and make predictions.
- <b>Interpretation:</b> The slope indicates the strength and direction of the relationship between the variables—whether they are positively or negatively correlated.

#### What is random state?

In the context of machine learning, particularly when splitting datasets, the random_state parameter is used to control the randomness of the data splitting process. Here's a brief explanation:

- <b>Purpose:</b> It ensures that the results are reproducible. When you set a specific value for random_state, you'll get the same split of the data every time you run your code. This is particularly useful for debugging and comparing models.
- <b>How it works:</b> The value of random_state acts as a seed for the random number generator. Using the same seed will yield the same random results.

For example, if you set random_state=2, every time you run the split, you'll get the same training and testing data. If you don't set it, you'll get different splits each time, which can lead to varying results.

---

#### Why is random state set to 2?

he choice of setting random_state to 2 (or any specific number) is somewhat arbitrary; it can be any integer. Here are a few reasons why you might see random_state=2 in code:

- <b>Reproducibility:</b> By setting it to 2, the author of the code ensures that anyone running the same code will get the same split of the dataset every time. This is useful for consistency in experiments.
- <b>Personal Preference:</b> Some developers choose specific numbers based on their own preference or simply pick a number they like. It has no inherent significance.
- <b>Testing and Validation:</b> Sometimes, developers will try different values of random_state to see how the model performs with different splits, but they may settle on a specific number for demonstration purposes.
    
Ultimately, you could set random_state to any integer (like 0, 1, 42, etc.), and it would serve the same purpose. The key is to keep it consistent for reproducibility.

#### What are the different methods through which we can find the value of m and b in linear regression?

In the context of mathematical modeling, particularly in optimization and regression, the terms "closed form" and "non-closed form" solutions refer to how the solution is expressed and derived. Here’s a breakdown of each:

- <b>Closed Form Solution</b>
    - <b>Definition:</b> A closed form solution is an explicit expression that directly computes the result using a finite number of standard operations (addition, subtraction, multiplication, division, exponentiation, etc.). It allows you to calculate the result in one step, without the need for iterative methods or approximations.
    - <b>Example in Linear Regression:</b> The normal equation for linear regression is a closed form solution: \[ \theta = (XTX){-1}X^Ty \] Here, \( \theta \) represents the coefficients (including \( m \) and \( b \)), and you can compute it directly from the data matrix \( X \) and the target vector \( y \).
    - <b>Advantages:</b> Closed form solutions are often faster and easier to implement for small to moderately sized datasets since they don't require iterations.
    - Method used is OLS(Ordinary Least Square)
    - Scikitlearn is used to implement OLS
    
- <b>Non-Closed Form Solution</b>
    - <b>Definition:</b> A non-closed form solution does not have a simple expression and often requires iterative methods or approximations to compute the result. These solutions may involve algorithms that converge towards the solution over multiple steps.
    - <b>Example in Linear Regression:</b> Gradient descent is a non-closed form solution. Instead of calculating \( m \) and \( b \) directly, it starts with initial guesses and iteratively updates them based on the gradient of the cost function until convergence is achieved.
    - <b>Advantages:</b> Non-closed form solutions can be more flexible and applicable to larger datasets or more complex models (e.g., when the features are numerous or when regularization techniques are used). They can also handle cases where closed forms do not exist.
    - Method used is Gradiant Descent
    - SGDRegressor is used to implement Gradient Descent
    
- <b>Summary</b>
    - <b>Closed Form:</b> Direct, explicit formula; quick computation; best for smaller datasets.
    - <b>Non-Closed Form:</b> Iterative process; approximates solutions; better for larger datasets or complex models.

#### If we have OLS method solve a Regression problem then why do we need Gradient Descent?

- <b>OLS (Ordinary Least Squares):</b>
    - <b>Closed-form Solution:</b> OLS provides a direct formula to calculate the coefficients. However, this approach involves computing the inverse of a matrix, which can become computationally expensive and unstable as the number of dimensions (features) increases.
    - <b>High Dimensionality Challenges:</b> In high dimensions, the matrix may be singular or near-singular, making it difficult to compute the inverse accurately.
- <b>Gradient Descent:</b>
    - <b>Iterative Approach:</b> Gradient Descent doesn't require the inversion of a matrix. Instead, it iteratively updates the coefficients based on the gradient of the loss function. This makes it more manageable for high-dimensional data.
    - <b>Handling Large Dimensions:</b> Gradient Descent can effectively operate with many features, as it processes data in batches, allowing for scalability without the need for all data to fit in memory.
In summary, while OLS works well with low to moderate dimensions, Gradient Descent shines in high-dimensional contexts where computational efficiency and flexibility are crucial.

### Regression Metrics

#### What is MAE, its advantages and disadvantages?

Mean Absolute Error (MAE) is a metric used to assess the accuracy of a regression model by calculating the average absolute differences between predicted and actual values. It measures how far off predictions are from the actual outcomes, without considering the direction of the errors.

- <b>Advantages of MAE:</b>
    - <b>Simplicity:</b> MAE is straightforward to understand and calculate, making it accessible for quick assessments of model performance.
    - <b>Same Units:</b> The errors are expressed in the same units as the target variable, which makes interpretation easier.
    - <b>Robust to Outliers:</b> MAE treats all errors equally, making it less sensitive to outliers compared to metrics like Mean Squared Error (MSE), which squares the errors.
    - <b>Linear Scale:</b> Unlike MSE, which emphasizes larger errors due to squaring, MAE provides a linear score that reflects the average error directly.
- <b>Disadvantages of MAE:</b>
    - <b>Less Sensitive to Error Magnitude:</b> Since MAE treats all errors equally, it may not adequately penalize larger errors, which can be important in certain applications.
    - <b>Non-Differentiable:</b> MAE is not differentiable at zero, which can make optimization more challenging for some algorithms that rely on gradient-based methods.
    - <b>Limited Information:</b> MAE provides only the average error and does not provide insights into the distribution of errors (e.g., whether there are many small errors or a few large ones).
- <b>Summary</b><br>
MAE is a valuable metric for evaluating regression models, especially when you need a simple and interpretable measure of prediction accuracy. However, depending on the specific context and goals of your analysis, you might also consider other metrics like MSE or R-squared for a more comprehensive evaluation.

#### What is MSE, its advantages and disadvantages?
 

Mean Squared Error (MSE) is a metric used to evaluate the performance of regression models by calculating the average of the squared differences between predicted and actual values. It measures how well a model performs in terms of prediction accuracy, giving more weight to larger errors due to squaring.

- <b>Formula for MSE:</b>
    - <b>MSE = frac{1} sum_^ (y_i - \hat_i)^2
    - Where:
    - n is the number of observations
    - y_i is the actual value
    - hat_i is the predicted value
- <b>Advantages of MSE:</b>
    - <b>Sensitivity to Large Errors:</b> Since MSE squares the errors, it gives more weight to larger errors. This can be beneficial when large errors are particularly undesirable in your application.
    - <b>Mathematical Properties:</b> MSE is differentiable, which is useful for optimization algorithms, especially those that rely on gradient descent.
    - <b>Useful for Theoretical Analysis:</b> MSE has well-established statistical properties and is commonly used in theoretical analyses, making it a standard choice in many contexts.
    - <b>Encourages Smaller Errors:</b> By penalizing larger deviations more heavily, MSE encourages models to reduce the larger errors, which can lead to better overall model performance in some cases.
- <b>Disadvantages of MSE:</b>
    - <b>Sensitivity to Outliers:</b> The squaring of errors means that MSE can be heavily influenced by outliers, potentially skewing the evaluation of model performance.
    - <b>Not in Original Units:</b> MSE is expressed in squared units of the target variable, which can make interpretation less intuitive compared to metrics like Mean Absolute Error (MAE).
    - <b>May Overemphasize Large Errors:</b> In some contexts, the emphasis on larger errors may not be desirable, especially if smaller errors are more relevant to the application.
- <b>Summary</b>
MSE is a widely used metric for evaluating regression models, particularly when larger errors are more critical and need to be addressed. However, it is essential to consider the context and potential drawbacks, such as sensitivity to outliers and the interpretation of squared units.

#### What is RMSE, its advantages and disadvantages?

RMSE, or Root Mean Square Error, is a metric used to measure the differences between predicted values by a model and the actual values. It’s calculated by taking the square root of the average of the squared differences between predicted and actual values.

- <b>Advantages of RMSE:</b>
    - <b>Sensitivity to Large Errors:</b> RMSE gives higher weight to larger errors, making it useful when large discrepancies are particularly undesirable.
    - <b>Easy Interpretation:</b> Since it's in the same units as the original data, RMSE is easy to interpret.
    - <b>Useful for Comparison:</b> It helps compare different models; a lower RMSE indicates a better fit.
- <b>Disadvantages of RMSE:</b>
    - <b>Outlier Sensitivity:</b> Because it squares the errors, RMSE can be disproportionately influenced by outliers.
    - <b>Complexity:</b> It may be less intuitive than other metrics like Mean Absolute Error (MAE) for some users.
    - <b>Not Scale-Invariant:</b> RMSE can be affected by the scale of the data, making it less suitable for comparing models across different datasets.

#### What is R Squared, its advantages and disadvantages?

R-squared (R²) is a statistical measure that indicates the proportion of the variance in the dependent variable that can be explained by the independent variable(s) in a regression model.
- <b>Formula:</b>
    - R^2 = 1 - SSR/SSM
    - Where:
        - SS_res (Sum of Squares of Residuals) = ∑(y_i - ŷ_i)² (the sum of the squares of the differences between actual values y_i and predicted values ŷ_i)
        - SS_tot (Total Sum of Squares) = ∑(y_i - ȳ)² (the sum of the squares of the differences between actual values y_i and the mean of actual values ȳ)
- <b>Intuition:</b>
    - <b>0 to 1 Scale:</b> R² ranges from 0 to 1. An R² of 0 means that the model explains none of the variability of the response data around its mean. An R² of 1 means that it explains all the variability.
    - <b>Explained Variance:</b> Essentially, R² tells you how much of the variability in the dependent variable can be explained by the independent variables. For example, an R² of 0.80 means that 80% of the variance is explained by the model.
- <b>Advantages of R-squared:</b>
    - <b>Simple Interpretation:</b> It provides a straightforward measure of how well the independent variables explain the variance in the dependent variable.
    - <b>Good for Model Comparison:</b> It allows for easy comparison between different models; a higher R² indicates a better fit.
    - <b>Widely Recognized:</b> R² is a standard metric in regression analysis, making it well-understood by many practitioners.
- <b>Disadvantages of R-squared:</b>
    - <b>Not Always Reliable:</b> A high R² does not necessarily mean that the model is good; it doesn’t indicate whether the model is appropriate or if it is overfitting.
    - <b>Ignores Model Complexity:</b> R² can increase with the addition of more variables, even if those variables do not contribute meaningfully to the model.
    - <b>Doesn't Indicate Causation:</b> A high R² does not imply a causal relationship between the variables; it merely indicates correlation.
    - <b>Sensitive to Outliers:</b> Outliers can disproportionately affect the R² value, leading to misleading interpretations.

---

#### What is the main flaw in R squared?
 

The main flaw in R-squared (R²) is that it can be misleading when assessing the quality of a regression model. Here are a few key points regarding this flaw:

- <b>Misleading Goodness-of-Fit:</b>
    - <b>High R² Doesn't Mean Good Model:</b> A high R² value does not necessarily indicate that the model is a good fit for the data. It can be artificially inflated by adding more predictors, even if those predictors do not have a meaningful relationship with the dependent variable. This leads to overfitting, where the model captures noise instead of the underlying pattern.
- <b>Other Limitations:</b>
    - <b>Ignores Model Complexity:</b> R² does not account for the number of predictors in the model. This means that a more complex model may have a higher R² simply due to having more parameters, not because it is a better model.
    - <b>Does Not Indicate Causation:</b> R² measures correlation, not causation. A high R² does not imply that changes in the independent variables cause changes in the dependent variable.
    - <b>Sensitivity to Outliers:</b> Outliers can significantly affect the R² value, leading to a distorted view of how well the model fits the data.

#### What is Adjusted R Squared, its formula, intution, advantages and disadvantages?

Adjusted R-squared (Adjusted R²) is a modified version of R-squared that adjusts for the number of predictors in a regression model. It provides a more accurate measure of the goodness-of-fit, especially when comparing models with different numbers of independent variables.
- <b>Formula:</b>
     - <b>Adj R2 = 1 - {(1 - R2)(n - 1)}/{n - p - 1}</b>
    - Where:
        -R² is the original R-squared value.
        -n is the number of observations (data points).
        -p is the number of predictors (independent variables) in the model.
- <b>Intuition:</b>
    - <b>Penalty for Complexity:</b> Adjusted R² adds a penalty for including additional predictors that do not improve the model significantly. This means that it will only increase if the new predictor improves the model more than would be expected by chance.
    - <b>Comparison Across Models:</b> It is particularly useful for comparing models with different numbers of predictors, as it provides a more reliable measure of how well the model generalizes to new data.
- <b>Advantages of Adjusted R-squared:</b>
    - <b>Accounts for Model Complexity:</b> It provides a more accurate measure of fit by considering the number of predictors, making it less likely to be misleading than R².
    - <b>Better for Model Comparison:</b> Adjusted R² can be used to compare models with different numbers of predictors, as it penalizes unnecessary complexity.
    - <b>Reduced Risk of Overfitting:</b> Since it adjusts for the number of predictors, it helps in identifying models that may be overfitting the data.
- <b>Disadvantages of Adjusted R-squared:</b>
    - <b>Less Intuitive:</b> The interpretation of Adjusted R² can be less straightforward compared to R², particularly for those unfamiliar with the concept.
    - <b>Still Doesn't Indicate Causation:</b> Like R², it does not imply causation; it merely indicates how well the independent variables explain the variance in the dependent variable.
    - <b>Can Decrease with High Noise:</b> If the model is poorly specified or contains a lot of noise, Adjusted R² can decrease even with the addition of new predictors.
    
Overall, Adjusted R² is a valuable metric when assessing the performance of regression models, especially when comparing models with different numbers of predictors

### Multiple Linear Regression

#### What is Multiple Linear Regression?

Multiple Linear Regression (MLR) is a statistical technique that models the relationship between one dependent variable and two or more independent variables. The goal is to predict the value of the dependent variable based on the values of the independent variables.
- <b>Key Points:</b>
    - <b>Equation:</b> The relationship is represented by the equation: Y = b_0 + b_1X_1 + b_2X_2 + ... + b_nX_n + epsilon Where:
        - <b>Y is the dependent variable (what you want to predict).
        - <b>b_0 is the intercept (value of \(Y\) when all \(X\) are 0).
        - <b>b_1, b_2, ..., b_n are the coefficients (show the impact of each independent variable).
        - <b>X_1, X_2, ..., X_n are the independent variables.
        - <b>epsilon is the error term (the difference between predicted and actual values).
    - <b>Assumptions:</b> MLR assumes a linear relationship, independence of errors, homoscedasticity (constant variance of errors), and normally distributed errors.
    - <b>Applications:</b> It's widely used in various fields such as economics, biology, engineering, and social sciences for predictions and forecasting.

---

#### What is the problem with OLS method?
 

- <b>Time Complexity:</b> As the number of variable increases in the data, the time to calculate inverse of matrix to get the value of beta increases. Hence the process is complex and time consuming.
- <b>Assumption of Linearity:</b> OLS assumes a linear relationship between the independent and dependent variables. If the actual relationship is non-linear, OLS may provide misleading results.
- <b>Sensitivity to Outliers:</b> OLS is sensitive to outliers, which can disproportionately affect the estimates of coefficients, leading to biased results.
- <b>Multicollinearity:</b> When independent variables are highly correlated, it can cause instability in the coefficient estimates. This makes it difficult to determine the individual effect of each variable.
- <b>Homoscedasticity Assumption:</b> OLS assumes that the variance of the errors is constant across all levels of the independent variables. If this assumption is violated (heteroscedasticity), it can lead to inefficient estimates and affect hypothesis testing.
- <b>Independence of Errors:</b> OLS assumes that the residuals (errors) are independent. If there's autocorrelation (common in time series data), it can lead to unreliable estimates.
- <b>Normality of Errors:</b> For valid hypothesis testing, OLS assumes that the residuals are normally distributed. If the errors are not normally distributed, it can affect the significance tests of the coefficients.
- <b>Overfitting:</b> If too many independent variables are included in the model, OLS may fit the noise in the data rather than the underlying relationship, leading to overfitting.
