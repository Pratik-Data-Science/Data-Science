### 3.0 Load the data

```python
# Load the diabetes dataset
# 'X' will contain the feature data, and 'y' will contain the target variable (diabetes progression)
X, y = load_diabetes(return_X_y=True)
```

```python
# Print the shape of the feature matrix 'X'
# This will show the number of samples (rows) and features (columns) in the dataset
print(X.shape)

# Print the shape of the target variable 'y'
# This will show the number of samples in the target variable (should match the number of rows in 'X')
print(y.shape)
```

### 4.0 MLR using Sklearn

```python
# Split the dataset into training and testing sets
# 80% of the data will be used for training and 20% for testing
# 'random_state' ensures reproducibility of the split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)

# Print the shape of the training feature matrix 'X_train'
# This shows the number of samples (rows) and features (columns) in the training set
print(X_train.shape)

# Print the shape of the testing feature matrix 'X_test'
# This shows the number of samples (rows) and features (columns) in the testing set
print(X_test.shape)

# Create an instance of the Linear Regression model
reg = LinearRegression()

# Fit the model to the training data
# This trains the model using 'X_train' as the input features and 'y_train' as the target variable
reg.fit(X_train, y_train)

# Predict the target variable for the test set using the trained model
# 'y_pred' will contain the predicted values for 'y_test'
y_pred = reg.predict(X_test)
```

```python
# Print the R² score (coefficient of determination) for the model
# This metric indicates how well the model explains the variability of the target variable
print("R2 Square:", r2_score(y_test, y_pred))

# Print the regression coefficients from the linear regression model 'reg'
# These coefficients represent the change in the target variable for a one-unit change in each predictor variable
print("Regression Coefficients:", reg.coef_)

# Print the regression intercept from the linear regression model 'reg'
# This intercept represents the expected value of the target variable when all predictor variables are set to zero
print("Regression Intercept:", reg.intercept_)
```

### 5.0 Making our own Linear Regression Class

```python
# Define a class for a simple Linear Regression model
class MeraLR:
 
    def __init__(self):
        # Initialize the coefficients and intercept as None
        self.coef_ = None
        self.intercept_ = None
 
    def fit(self, X_train, y_train):
        # Add a column of ones to the feature matrix to account for the intercept
        X_train = np.insert(X_train, 0, 1, axis=1)

        # Calculate the coefficients using the Normal Equation
        # betas = (X^T * X)^(-1) * (X^T * y)
        betas = np.linalg.inv(np.dot(X_train.T, X_train)).dot(X_train.T).dot(y_train)
        
        # The first coefficient is the intercept
        self.intercept_ = betas[0]
        
        # The rest of the coefficients correspond to the features
        self.coef_ = betas[1:]
 
    def predict(self, X_test):
        # Calculate predictions for the test set
        # y_pred = X * coef + intercept
        y_pred = np.dot(X_test, self.coef_) + self.intercept_
        return y_pred
```

```python
# Create an instance of the MeraLR class (our custom linear regression model)
lr = MeraLR()

# Fit the model to the training data
# This will calculate the coefficients and intercept based on 'X_train' and 'y_train'
lr.fit(X_train, y_train)

# Print the shape of the training feature matrix 'X_train'
# This shows the number of samples (rows) and features (columns) in the training set
print(X_train.shape)

# Print the shape of the training feature matrix after inserting a column of ones
# This is to see how the shape changes when we add the intercept term
print(np.insert(X_train, 0, 1, axis=1).shape)

# Use the fitted model to predict the target variable for the test set
# 'y_pred' will contain the predicted values for 'y_test'
y_pred = lr.predict(X_test)
```

```python
# Print the R² score (coefficient of determination) for the model
# This metric indicates how well the model explains the variability of the target variable
print("R2 Square:", r2_score(y_test, y_pred))

# Print the regression coefficients from the linear regression model 'reg'
# These coefficients represent the change in the target variable for a one-unit change in each predictor variable
print("Regression Coefficients:", lr.coef_)

# Print the regression intercept from the linear regression model 'reg'
# This intercept represents the expected value of the target variable when all predictor variables are set to zero
print("Regression Intercept:", lr.intercept_)
```
