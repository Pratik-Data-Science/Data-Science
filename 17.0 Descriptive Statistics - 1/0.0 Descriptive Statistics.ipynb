### Descriptive Statistics - Theory

#### What is Statistics?

Statistics is a branch of mathematics that involves collecting, 
analysing, interpreting, and presenting data. It provides tools and 
methods to understand and make sense of large amounts of data 
and to draw conclusions and make decisions based on the data.

In practice, statistics is used in a wide range of fields, such as 
business, economics, social sciences, medicine, and engineering. It is 
used to conduct research studies, analyse market trends, evaluate 
the effectiveness of treatments and interventions, and make 
forecasts and predictions.

Examples:
- <b>Business:</b> Data Analysis(Identifying customer behavior) and Demand Forecasting
- <b>Medical:</b> Identify efficacy of new medicines(Clinical trials), Identifying risk factor for diseases(Epidemiology)
- <b>Government & Politics:</b> Conducting surveys, Polling
- <b>Environmental Science:</b> Climate research

There are majorly 2 types of statistics:

<b>1. Descriptive Statistics:</b> Descriptive statistics deals with the 
collection, organization, analysis,

It focuses on summarizing and describing

#### What is population and sample?

<b>- Population</b> refers to the entire group of individuals or objects that we are 
interested in studying. It is the complete set of observations that we want to make 
inferences about. For example, the population might be all the students in a 
particular school or all the cars in a particular city.

<b>- Sample</b>, on the other hand, is a subset of the population. It is a smaller group of 
individuals or objects that we select from the population to study. Samples are 
used to estimate characteristics of the population, such as the mean or the 
proportion with a certain attribute. For example, we might randomly select 100 
students.

Examples
1. All cricket fans vs fans who were present in the stadium
2. All students vs who visit college for lectures

Things to be careful about while creating samples
1. Sample Size
2. Random
3. Representative

#### What is descriptive statistics?

Descriptive statistics involves the use of numerical and graphical methods to summarize and describe the important features of a dataset. It helps in providing simple summaries about the sample and the measures. Descriptive statistics are used to describe the basic features of the data in a study. They provide simple summaries about the sample and the measures. These summaries may either form the basis of the initial description of the data as part of a more extensive statistical analysis, or they may be sufficient in and of themselves for certain purposes.

Descriptive statistics commonly include measures such as mean, median, mode, standard deviation, range, and percentiles. These metrics help to give a clear and concise understanding of the characteristics of the data, such as central tendency, dispersion, and the shape of the distribution. They are valuable tools for summarizing and interpreting large datasets.

---

#### What is inferential statistics?

Inferential statistics is a branch of statistics that deals with making inferences or predictions 
about a larger population based on a sample of data. It involves using statistical techniques to 
test hypotheses and draw conclusions from data. Some of the topics that come under 
inferential statistics are:

<b>1. Hypothesis testing:</b> This involves testing a hypothesis about a population parameter based 
on a sample of data. For example, testing whether the mean height of a population is 
different from a given value.

<b>2. Confidence intervals:</b> This involves estimating the range of values that a population 
parameter could take based on a sample of data. For example, estimating the population 
mean height within a given confidence level.

<b>3. Analysis of variance (ANOVA):</b> This involves comparing means across multiple groups to 
determine if there are any significant differences. For example, comparing the mean 
height of individuals from different regions.

<b>4. Regression analysis:</b> This involves modelling the relationship between a dependent 
variable and one or more independent variables. For example, predicting the sales of a 
product based on advertising expenditure.

<b>5. Chi-square tests:</b> This involves testing the independence or association between two 
categorical variables. For example, testing whether gender and occupation are 
independent variables.

<b>6. Sampling techniques:</b> This involves ensuring that the sample of data is representative of 
the population. For example, using random sampling to select individuals from a 
population.

<b>7. Bayesian statistics:</b> This is an alternative approach to statistical inference that involves 
updating beliefs about the probability of an event based on new evidence. For example, 
updating the probability of a disease given a positive test result.

#### What is Parameter and Statistic?

A parameter is a characteristic of a population, while a statistic is a characteristic 
of a sample. Parameters are generally unknown and are estimated using statistics. 
The goal of statistical inference is to use the information obtained from the sample 
to make inferences about the population parameters.

#### What are different types of data?

There are majorly 2 types of data

<b>1. Categorical or Qualitative data: </b> Categorical data is a type of data that represents characteristics or qualities. It uses labels or categories to describe attributes of a population. This type of data is often represented by non-numeric values and is used to group or categorize information.
- <b>Nominal Data:</b> This type of data represents categories without any inherent order or ranking. Examples include colors, names, or categories of products.
- <b>Ordinal Data:</b> Ordinal data represents categories with a natural order, but the differences between the categories are not precisely measurable. Examples include rankings (e.g., 1st, 2nd, 3rd) or rating scales (e.g., low, medium, high).

<b>2. Numerical or Quantitative data: </b>Numerical data, also known as quantitative data, represents measurable quantities and is expressed in numbers. This type of data is used for performing mathematical calculations and statistical analysis. Numerical data can be further categorized into discrete and continuous data:
- <b>Discrete Data:</b> Discrete data can only take specific values within a certain range. These values are typically whole numbers and cannot be broken down into smaller parts. Examples include the number of students in a class, the number of cars in a parking lot, or the number of items sold.
- <b>Continuous Data:</b> Continuous data can take any value within a certain range and can be measured at any level of precision. Examples include height, weight, temperature, and time.

#### What is measure of Central Tendency?

A measure of central tendency is a statistical measure that represents a typical or 
central value for a dataset. It provides a summary of the data by identifying a 
single value that is most representative of the dataset as a whole. 

---

#### What are the measures of Central Tendency?

<b>1. Mean:</b> The mean is the sum of all values in the dataset divided by the number of 
values. It is generally prone to outliers.<br>
<b>2. Median:</b> The median is the middle value in the dataset when the data is arranged 
in order.<br>
<b>3. Mode:</b> The mode is the value that appears most frequently in the dataset. It is used on categorical columns.<br>
<b>4. Weighted Mean:</b> The weighted mean is the sum of the products of each value and 
its weight, divided by the sum of the weights. It is used to calculate a mean when 
the values in the dataset have different importance or frequency. <br>
<b>5. Trimmed Mean:</b> A trimmed mean is calculated by removing a certain percentage of the 
smallest and largest values from the dataset and then taking the mean 
of the remaining values. The percentage of values removed is called the 
trimming percentage.

#### What is population and sample mean?

The population mean and sample mean are both measures of central tendency that indicate the average value of a dataset. Here's a brief explanation of each:

<b>- Population Mean:</b> The population mean represents the average value of a variable in the entire population. It is calculated by summing up all the values in the population and then dividing by the total number of values. The population mean is denoted by the Greek letter "μ" (mu).
<b>μ = (Σx) / N</b> Where: μ = Population mean Σx = Sum of all values in the population N = Total number of values in the population<br>
<b>- Sample Mean:</b> The sample mean represents the average value of a variable in a sample of the population. It is calculated by summing up all the values in the sample and then dividing by the total number of values in the sample. The sample mean is denoted by the symbol "x̄" (x-bar).
<b>(Σx) / n</b> Where: x̄ = Sample mean Σx = Sum of all values in the sample n = Total number of values in the sample<br>

It's important to note that the sample mean is an estimate of the population mean. When working with statistical analysis, researchers often use sample means to make inferences about population means due to the practical challenges of collecting data from entire populations.

#### What is measure of Disperssion?

A measure of dispersion is a statistical measure that describes the spread or 
variability of a dataset. It provides information about how the data is distributed 
around the central tendency (mean, median or mode) of the dataset.

---

#### What are the measures of Disperssion?

<b>1. Range:</b> The range is the difference between the maximum and minimum values in 
the dataset. It is a simple measure of dispersion that is easy to calculate but can be 
affected by outliers. Range = Maximum Value - Minimum Value<br>
<b>2. Variance:</b>  The variance is the average of the squared differences between each 
data point and the mean. It measures the average distance of each data point 
from the mean and is useful in comparing the dispersion of datasets with different 
means. σ2 = Σ(xi - μ)^2 / N<br>
<b>3. Standard Deviation:</b> The standard deviation is the square root of the variance. It is 
a widely used measure of dispersion that is useful in describing the shape of a 
distribution. σ = √(Σ(xi - μ)^2 / N)<br>
<b>4. Coefficient of Variation (CV):</b> The CV is the ratio of the standard deviation to the 
mean expressed as a percentage. It is used to compare the variability of datasets 
with different means and is commonly used in fields such as biology, chemistry, 
and engineering.
The coefficient of variation (CV) is a statistical measure that expresses the amount 
of variability in a dataset relative to the mean. It is a dimensionless quantity that is 
expressed as a percentage.
The formula for calculating the coefficient of variation is:
CV = (standard deviation / mean) x 100%<br>

#### What is mean absolute deviation?

Mean Absolute Deviation (MAD) is a measure of variability that gives the average distance between each data point and the mean. It's calculated by taking the absolute differences between each data point and the mean, then finding the average of those absolute differences. The formula for MAD is: MAD = Σ|xi - μ| / N

#### What is population and sample variance?

Population variance and sample variance are measures of the spread of a set of data points around the mean.

<b>Population Variance (σ^2):</b><br>
It measures the variability of the entire population.
The formula is: σ^2 = Σ(xi - μ)2 / N, where σ2 is the population variance, Σ is the sum of, xi is each individual value, μ is the mean of the values, and N is the total number of values.

<b>Sample Variance (s^2):</b><br>
It estimates the variability of a sample from a larger population.
The formula is: s^2 = Σ(xi - x̄)2 / (n-1), where s2 is the sample variance, Σ is the sum of, xi is each individual value, x̄ is the sample mean, and n is the sample size.

Remember, the key difference is in the denominator: N for population variance and (n-1) for sample variance.

---

#### Why is the denominator of sample variance n-1?

The denominator of the sample variance formula is (n-1) instead of n to account for the degrees of freedom in the sample. Degrees of freedom represent the number of values in the final calculation of a statistic that are free to vary.

When estimating the population variance from a sample, using (n-1) instead of n as the denominator provides an unbiased estimate of the population variance. This adjustment is necessary because the sample mean is used in the calculation, and it tends to be closer to the data points than the population mean. As a result, using n-1 in the denominator helps to account for this issue and provides a more accurate estimation of the population variance.

This adjustment is based on statistical principles and ensures that the sample variance provides a better representation of the true variability in the population from which the sample was drawn. I hope this clarifies the reason behind using (n-1) in the denominator of the sample variance formula!

#### Graphs for Univariate analysis?

<b>1. Categorical column:</b><br>
<b>i. Frequency Distribution Table:</b> A frequency distribution table is a table that summarizes the number of times (or 
frequency) that each value occurs in a dataset. We can plot bar chart<br>
<b>ii. Relative frequency:</b> Relative frequency is the proportion or percentage of a category in a dataset or 
sample. It is calculated by dividing the frequency of a category by the total number 
of observations in the dataset or sample. We can plot pie chart.<br>
<b>iii Cummlative frequency:</b> Cumulative frequency is the running total of frequencies of a variable or category 
in a dataset or sample. It is calculated by adding up the frequencies of the current 
category and all previous categories in the dataset or sample. We can plot line chart.

<b>2. Numerical column:</b>
Histogram can be used to plot the numerical column once it is converted ti bins. Based on the shape of chart, histogram can be of diffrent types<br>
<b>i. Symmetric Histogram:</b> In a symmetric histogram, the data is evenly distributed around the mean, resulting in a bell-shaped curve when plotted. This indicates that the data has a symmetrical distribution.<br>
<b>ii. Bimodal Histogram:</b> A bimodal histogram has two distinct peaks, indicating that the data has two modes or prominent peaks in its distribution.<br>
<b>iii. Left-Skewed Histogram:</b> Also known as negatively skewed, a left-skewed histogram has a long tail on the left side, with the majority of data points concentrated on the right side. The mean is typically less than the median in a left-skewed distribution.<br>
<b>iv. Right-Skewed Histogram:</b> Also known as positively skewed, a right-skewed histogram has a long tail on the right side, with the majority of data points concentrated on the left side. The mean is typically greater than the median in a right-skewed distribution.<br>
<b>v. Uniform (No Pattern) Histogram:</b> In a uniform histogram, all bars have approximately the same height, indicating that the data is evenly distributed across the range of values. There is no apparent peak or skewness in the distribution.
